# Getting Started with Neurodesktop

Welcome to **Neurodesktop** (built on **scidesktop**). This notebook summarizes the core tools available in this session:

- **Neuroimaging tools (Lmod + CVMFS / sciget):** run `module avail` in a terminal or use the **Software Modules** sidebar; when CVMFS is not mounted, install containers on demand with `sciget find <tool>` and `sciget install <tool>/<version>`.
- **Neuroimaging Python kernel:** switch to the **Neuroimaging (Python 3.13)** kernel for `nibabel`, `nipype`, `niwrap`, `pydra`, `nipoppy`, `ipyniivue`, `datalad`, `matplotlib`, and `scipy`.
- **Image viewers:** double-click any `.nii` / `.nii.gz` file in the file browser to open **NiiVue**, or right-click → **Open With → FreeBrowse**.
- **Workflows & Batch jobs:** `nextflow`, `nf-test`, `snakemake`, `datalad`, and a local **Slurm** queue (`sbatch`, `srun`, and the JupyterLab Slurm panel) are ready out of the box.
- **Citations & Helpers:** run `scigethelpers cite` to generate `CITATIONS.md` for every loaded container module, `scigethelpers info`, `scigethelpers check`, or `scigethelpers kernel add <module>`.
- **AI & Editors:** open **VS Code**, **scigent.ai (T3 Code)**, or **Chat (Jupyter AI)** from the Launcher, or use `codex`, `claude`, and `opencode` in a terminal.
